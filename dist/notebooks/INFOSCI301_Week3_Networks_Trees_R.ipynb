## 1. Verify the same graph contract

```r
labels <- c('Claim','Paper','Code','Dataset','License','Model','Evaluation','User task')
nodes <- LETTERS[1:8]
edges <- matrix(c('A','B','A','G','A','H','B','C','B','D','C','D','C','F','D','E','D','F','F','G','G','H'),ncol=2,byrow=TRUE)
A <- matrix(0L,8,8,dimnames=list(nodes,nodes))
for(i in seq_len(nrow(edges))) { a <- edges[i,1]; b <- edges[i,2]; stopifnot(a != b); A[a,b] <- 1L; A[b,a] <- 1L }
stopifnot(all(A==t(A)), all(diag(A)==0), sum(A)==2*nrow(edges))
print(c(nodes=length(nodes),edges=nrow(edges)))
print(rowSums(A))
```

## 2. Compare the same tasks in two idioms

```r
op <- par(mfrow=c(1,2),mar=c(3,3,3,1))
theta <- (0:7)*2*pi/8
xy <- cbind(cos(theta),sin(theta)); rownames(xy) <- nodes
plot(xy,type='n',axes=FALSE,xlab='',ylab='',main='Trace A–B–D',asp=1,xlim=c(-1.2,1.2),ylim=c(-1.2,1.2))
for(i in seq_len(nrow(edges))) { a <- xy[edges[i,1],]; b <- xy[edges[i,2],]; segments(a[1],a[2],b[1],b[2],col='#9fb3c8') }
points(xy,pch=21,bg='#2563eb',col='white',cex=4);text(xy,labels=nodes,col='white')
plot(0,0,type='n',xlim=c(.5,8.5),ylim=c(8.5,.5),axes=FALSE,xlab='',ylab='',main='Compare D and F',asp=1)
for(i in 1:8) for(j in 1:8) { rect(j-.48,i-.48,j+.48,i+.48,col=if(A[i,j]) '#2563eb' else '#edf2f7',border=NA);text(j,i,A[i,j],col=if(A[i,j]) 'white' else '#52667a') }
axis(3,at=1:8,labels=nodes,tick=FALSE);axis(2,at=1:8,labels=nodes,tick=FALSE,las=1)
par(op)
print(intersect(nodes[A['D',]==1],nodes[A['F',]==1]))
```

## 3. Derive a breadth-first tree and report what it omits

```r
seen <- 'A'; queue <- 'A'; tree <- matrix(character(),ncol=2)
while(length(queue)>0) {
  a <- queue[1]; queue <- queue[-1]
  for(b in nodes[A[a,]==1]) if(!(b %in% seen)) {
    seen <- c(seen,b); queue <- c(queue,b); tree <- rbind(tree,c(a,b))
  }
}
stopifnot(setequal(seen,nodes),nrow(tree)==length(nodes)-1)
print(tree)
key <- function(m) apply(m,1,function(x) paste(sort(x),collapse='-'))
print(setdiff(key(edges),key(tree)))
```
